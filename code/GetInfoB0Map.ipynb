```python
import subprocess

def afni(cmd):
    out = subprocess.run(cmd, check=True, text=True, capture_output=True)
    return out.stdout.strip()

dset = "SpinEchoFieldMap_AP.nii.gz"

pixel  = list(map(int,   afni(["3dinfo", "-n3",     dset]).split()))
voxdim = list(map(float, afni(["3dinfo", "-ad3",    dset]).split()))
origin = list(map(float, afni(["3dinfo", "-o3",     dset]).split()))
extent = list(map(float, afni(["3dinfo", "-extent", dset]).split()))
orient = afni(["3dinfo", "-orient", dset])
nvol   = int(afni(["3dinfo", "-nv", dset]))

print("pixel =", pixel)      # [ni, nj, nk]
print("voxdim =", voxdim)    # [dx, dy, dz] in mm
print("origin =", origin)    # dataset i/j/k origin
print("extent =", extent)    # [R, L, A, P, I, S]
print("orient =", orient)    # e.g. LPI
print("nvol =", nvol)        # number of sub-bricks
```
