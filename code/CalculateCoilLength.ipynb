```python
import math

# 常数 R（可根据需要修改）
R = 130
# 制作线圈时使用的，不参与电流控制
def compute_l(r, R=R):
    """
    根据公式 l = 2R * arcsin(r/R) 计算 l
    
    参数:
        r (float): 输入值
        R (float): 常数 R
    
    返回:
        float: 计算得到的 l
    """
    if abs(r / R) > 1:
        raise ValueError("Invalid input: |r/R| must be <= 1")
    
    return 2 * R * math.asin(r / R)


if __name__ == "__main__":
    try:
        r = float(input("请输入 r 的值: "))
        l = compute_l(r)
        print(f"计算得到的 l = {l}")
    except Exception as e:
        print(f"错误: {e}")
```
